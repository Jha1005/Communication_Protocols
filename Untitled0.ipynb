```python
import os
import yaml
from ultralytics import YOLO

# Use the DOWNLOAD_DIR defined in the previous cell
dataset_root = DOWNLOAD_DIR

print(f"Dataset root location: {dataset_root}")
print("Contents of dataset_root:")
!ls -R {dataset_root}

data_yaml_path = None

# Option 1: Check directly in dataset_root
if os.path.exists(os.path.join(dataset_root, "data.yaml")):
    data_yaml_path = os.path.join(dataset_root, "data.yaml")
else:
    # Option 2: Look for data.yaml in subdirectories (common for Roboflow exports)
    for root, dirs, files in os.walk(dataset_root):
        if "data.yaml" in files:
            data_yaml_path = os.path.join(root, "data.yaml")
            break

if data_yaml_path is None:
    raise FileNotFoundError(f"data.yaml not found in {dataset_root} or its subdirectories.")

print(f"Resolved data.yaml path for training: {data_yaml_path}")

# Load the data.yaml file
with open(data_yaml_path, 'r') as f:
    data_config = yaml.safe_load(f)

# Modify the 'val' path if it points to a non-existent directory
if 'val' in data_config and not os.path.exists(os.path.join(dataset_root, data_config['val'])):
    print(f"Original 'val' path in data.yaml: {data_config['val']}")
    # Assuming 'train' directory exists and can be used for validation too if no 'valid' is present
    data_config['val'] = data_config['train'] # Use train path for validation
    print(f"Modified 'val' path in data.yaml to: {data_config['val']}")

    # Save the modified data.yaml back
    with open(data_yaml_path, 'w') as f:
        yaml.safe_dump(data_config, f)
    print(f"data.yaml updated at {data_yaml_path} with new 'val' path.")

print(f"\nContents of {data_yaml_path} after modification:")
!cat {data_yaml_path}

model = YOLO("yolov8n.pt")   # nano model
model.train(
    data=data_yaml_path,
    epochs=100,
    imgsz=640,
    batch=16
)
```
